The following code below creates three unique folders for each binary image classification class. These folders will include the train, test, and validation partitions for both the 'person neutral expression' and 'person smiling expression' categories of the person images. These will be utilized in the next steps for model fitment!

```
import os
import glob
import random
import shutil

# Define initial base directories and person neutral expression/person smiling expression class names
base_directory = 'images'
class_names = ['person neutral expression', 'person smiling expression']

# Create main directories if they don't exist
for dataset_type in ['train', 'test', 'validation']:
    for class_name in class_names:
        path = os.path.join(base_directory, dataset_type, class_name.replace(' ', '_')) # Replace spaces for cleaner folder names
        os.makedirs(path, exist_ok=True)
        print(f"Created directory: {path}")
```

The following code below splits up the images for each class into the train, test, and validation partitions according to a 60/20/20 split. Each image is then moved into the appropriate location in the train, test, and validation partitions located within the `person neutral expression` and `person smiling expression` folders.

```
# Assign the train, test, and validation partition splits according to 60/20/20
train_partition = 0.6
test_partition = 0.2
val_partition = 0.2

# This function processes each of the two classes by moving the designated Class 0 and Class 1 images into the
# train, test, and validation partitions for each class after randomly shuffling the images for optimal model learning

for class_name in class_names:
    print(f"\nProcessing class: '{class_name}'")

    # Construct the correct source path for the current class
    source_path = os.path.join('images', class_name)
    all_images = [os.path.join(source_path, f) for f in os.listdir(source_path) if os.path.isfile(os.path.join(source_path, f))]

    # Crucial step for ensuring all images are randomy shuffled!
    random.shuffle(all_images)

    # Obtain the necessary image counts for the train and test partitions
    # num_val is excluded as thetre is no need for explicitly defining it
    num_images = len(all_images)
    num_train = int(train_partition * num_images)
    num_test = int(test_partition * num_images)

    # Seelct the images for each parition
    train_images = all_images[:num_train]
    test_images = all_images[num_train : num_train + num_test]
    validation_images = all_images[num_train + num_test:]

    # Create the destination paths for current class using the original base_directory ('images')
    train_dest = os.path.join('images', 'train', class_name.replace(' ', '_'))
    test_dest = os.path.join('images', 'test', class_name.replace(' ', '_'))
    val_dest = os.path.join('images', 'validation', class_name.replace(' ', '_'))

    # Provide visual feedback to user regarding the train partitioning progress
    print(f"  Moving {len(train_images)} images to {train_dest}")
    for f in train_images:
        shutil.copy(f, train_dest)

    # Provide visual feedback to user regarding the train partitioning progress
    print(f"  Moving {len(test_images)} images to {test_dest}")
    for f in test_images:
        shutil.copy(f, test_dest)

    # Provide visual feedback to user regarding the train partitioning progress
    print(f"  Moving {len(validation_images)} images to {val_dest}")
    for f in validation_images:
        shutil.copy(f, val_dest)

print("\nTrain, test, and validation directories have been created. Images reshuffled and moved to appropriate subfolders for partitioning.")
```

The following code below offers a simple image count to verify that the downloaded images for both classes were split accoringly into their unique, designated partitions.

```
print("\nVerification of image counts:\n")
for dataset_type in ['train', 'test', 'validation']:
    print(f"{dataset_type.capitalize()}")
    for class_name in class_names:
        target_path = os.path.join('images', dataset_type, class_name.replace(' ', '_'))
        count = len([name for name in os.listdir(target_path) if os.path.isfile(os.path.join(target_path, name))])
        print(f"{class_name}: {count} images located.")
    print("\n")
```

**Task 3:** Fit a ConvNet model.

### Defining Paths, Creating ImageDataGenerators, and Feeding Data to IDGs

```
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Define the image dimensions, batch size, and number of classes (neutral, smiling)
img_height, img_width = 150, 150
batch_size = 32
num_classes = 2

# Define the paths to my data directories
train_dir = 'images/train'
validation_dir = 'images/validation'
test_dir = 'images/test'

# Create ImageDataGenerator for training (w/ data augmentation)
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Create ImageDataGenerator for validation and test with rescaling only
validation_test_datagen = ImageDataGenerator(rescale=1./255)

# Supply training images in batches w/ train_datagen
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='binary'
)

# Supply training images in batches w/ validation_test_datagen
validation_generator = validation_test_datagen.flow_from_directory(
    validation_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='binary'
)

# Supply training images in batches w/ validation_test_datagen
test_generator = validation_test_datagen.flow_from_directory(
    test_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='binary',
    shuffle=False
)
```

### Building the ConvNet Model

The following code below defines a simple Sequential CNN model with a few convolutional and pooling layers, followed by dense layers. The output layer uses a sigmoid activation function for binary classification.

```
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(img_height, img_width, 3)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(1, activation='sigmoid')
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

### Training the Model

The following code below trains the model using the `train_generator` and validates it with the `validation_generator`.

```
epochs = 30 # Selected after numerous tests, confirmed to perform well!
history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples // batch_size,
    epochs=epochs,
    validation_data=validation_generator,
    validation_steps=validation_generator.samples // batch_size
)
```

**Task 4:** Evaluate the ConvNet model.

### Model Evaluation
The following code below displays the test loss and test accuracy for the model.

```
print("Evaluating model on test data...")
eval_results = model.evaluate(
    test_generator,
    steps=test_generator.samples // batch_size
)

print(f"\nTest Loss: {eval_results[0]:.4f}")
print(f"Test Accuracy: {eval_results[1]:.4f}")
```

The following code below generates predictions and extracts true labels.

```
import numpy as np

# Generate predictions on the test set
y_pred_probs = model.predict(test_generator)

# Convert probabilities to binary class labels (0 or 1)
y_pred = np.round(y_pred_probs).flatten()

# Get true labels from the test_generator
y_true = test_generator.labels

print(f"Shape of predicted probabilities: {y_pred_probs.shape}")
print(f"Shape of binary predictions: {y_pred.shape}")
print(f"Shape of actual labels: {y_true.shape}")
```

The following code below generates the classification report for my model.



```
from sklearn.metrics import classification_report
print(f"Class indices: {train_generator.class_indices}")
target_names = ['person neutral expression', 'person smiling expression']

# Generate and print the classification report
report = classification_report(y_true, y_pred, target_names=target_names)
print("\nClassification Report:\n")
print(report)
```

The following code below generates the confusion matrix for my model.


```
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Compute the confusion matrix and prepare for display
cm = confusion_matrix(y_true, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=target_names)

# Display the confusion matrix plot
fig, ax = plt.subplots(figsize=(8, 8))
disp.plot(cmap=plt.cm.Blues, ax=ax)
ax.set_title('Confusion Matrix')
plt.show()
```
